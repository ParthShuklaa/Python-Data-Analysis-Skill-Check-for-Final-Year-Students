```python
import time

n = 5_000_000
a = np.arange(n)

# Loop version
start = time.time()
result = [x * 2 for x in a]
print("Loop:", time.time() - start)

# Vectorized version
start = time.time()
result = a * 2
print("Vectorized:", time.time() - start)
```
